# How to find 7tv emotes


global and channel specific emotes can be downloaded for cacheing unlike Twitch Emotes.

add this later for expansion

You will need to get an API for 7tv

```python
import requests
import os
from dotenv import load_dotenv
load_dotenv()

target_UserID = os.getenv("TWITCH_USER_ID")
```

```python
def get_7tv_emote_set(user_id):
    url = f"https://7tv.io/v3/users/twitch/{user_id}"
    data = requests.get(url).json()
    emote_set = {}

    if "emote_set" in data:
        emotes = data["emote_set"]["emotes"]
        for emote in emotes:
            emote_name = emote["name"]
            emote_id = emote["id"]
            emote_url = f"https://cdn.7tv.app/emote/{emote_id}/4x.webp"
            emote_set[emote_name] = emote_url
    else:
        print("No 7TV emote set found")

    return emote_set

_7tv_emote_set = get_7tv_emote_set(target_UserID)
_7tv_emote_set
```

# 7TV global emotes

```python
def update_emotes_global():
    # global emote sets
    url = f"https://7tv.io/v3/emote-sets/global"
    data = requests.get(url).json()
    emote_set = {}

    emotes = data["emotes"]
    for emote in emotes:
        emote_name = emote["name"]
        emote_id = emote["id"]
        emote_url = f"https://cdn.7tv.app/emote/{emote_id}/4x.webp"
        emote_set[emote_name] = emote_url

    if not emote_set:
        print("No Global 7TV emotes set found")

    return emote_set

update_emotes_global()
```

# BTTV Channel Emotes

```python
# use the user id, to get BetterTwitchTV emotes
def get_BTV_emote_set(user_id):
    url = f"https://api.betterttv.net/3/cached/users/twitch/{user_id}"
    data = requests.get(url).json()
    emote_set = {}

    types = ["sharedEmotes", "chennelEmotes"]
    for type in types:
        print("getting BTV emote set", type)

        emotes = data.get(type)
        if emotes:
            for emote in emotes:
                emote_id = emote["id"]
                emote_name = emote["code"]
                emote_url = f"https://cdn.betterttv.net/emote/{emote_id}/3x.webp"
                emote_set[emote_name] = emote_url

    if not emote_set:
        print("No BTV emote set found")

    return emote_set

# using cweamcat's for not since I don't have any BTTV emotes yet
BTV_emote_set = get_BTV_emote_set(262365865)
BTV_emote_set
```

# BTV Global Emotes

```python
# use the user id, to get BetterTwitchTV emotes
def get_BTV_global_emote_set():
    url = f"https://api.betterttv.net/3/cached/emotes/global"
    data = requests.get(url).json()
    emote_set = {}
    emotes = data

    if emotes:
        for emote in emotes:
            emote_id = emote["id"]
            emote_name = emote["code"]
            emote_url = f"https://cdn.betterttv.net/emote/{emote_id}/3x.webp"
            emote_set[emote_name] = emote_url

    if not emote_set:
        print("No BTV emote set found")

    return emote_set

BTV_emote_set = get_BTV_global_emote_set()
BTV_emote_set
```

# FFZ Channel Emotes

```python
# use the user id, to get FrankerFaceZ emotes
def get_FFZ_emote_set(user_id):
    r = requests.get(f"https://api.frankerfacez.com/v1/room/id/{user_id}")
    r.raise_for_status()
    data = r.json()

    emote_set = {}

    set_id = data["room"]["set"]
    emotes = data["sets"][str(set_id)]["emoticons"]

    for emote in emotes:
        emote_name = emote["name"]
        emote_url = emote["urls"]["4"]
        emote_set[emote_name] = emote_url

    if not emote_set:
        print("No FFZ emote set found")

    return emote_set


ffz_emote_set = get_FFZ_emote_set(262365865)
ffz_emote_set
```
