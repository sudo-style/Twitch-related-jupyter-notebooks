```python
import requests
from dotenv import load_dotenv
load_dotenv()
import os
```

```python
channel = "crelly"
year = 2026
month = 9
day = 5
limit = 100

res = requests.get(f"https://logxx.dev/channel/{channel}/{year}/{month}/{day}?json=&limit={limit}")
data = res.json()
for i in range(len(data['messages'][:10])):
    message = data['messages'][i]["text"]
    username = data['messages'][i]["username"]
    print(f"{i}: {username}: {message}")
```

# Get user logs of a channel in a month

```python
channel = "crelly"
username = "sudostyle"
year = 2026
month = 9

res = requests.get(f"https://logxx.dev/channel/{channel}/user/{username}/{year}/{month}?json=")
data = res.json()
for i in range(len(data['messages'][:10])):
    message = data['messages'][i]["text"]
    print(f"{i}: {message}")
```

# Gets a random user message

```python
channel = "crelly"
username = "sudostyle"
res = requests.get(f"https://logxx.dev/channel/{channel}/user/{username}/random?json=")
data = res.json()
message = data["messages"][0]
message
```

```python
text = message["text"]
user = message["displayName"]
time = message["timestamp"]

print(f"{time} {user}: {text}")
```

# Check Logs for a specific message

```python
channel = "crelly"
username = "sudostyle"
search_querry = "hello"
res = requests.get(f"https://logxx.dev/channel/{channel}/user/{username}/search?q={search_querry}&json=")
data = res.json()

for i in range(len(data['messages'])):
    message = data['messages'][i]["text"]
    print(f"{i}: {message}")
```
