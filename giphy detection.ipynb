# Geting the metadata

To get the metadata, you will need to make an API key with giphy, by making a [giphy account](https://developers.giphy.com).


```python
from dotenv import load_dotenv
load_dotenv()
import os
import requests

API_KEY = os.getenv("GIPHY_API_KEY")
API = f"https://api.giphy.com/v1/gifs/{gif_id}?api_key={API_KEY}"
response = requests.get(API)
response
```

```python
data = response.json()["data"]
print(f"username: {data.get("username")}")
print(f"source: {data.get("source")}")
print(f"GIF URL: {data.get("url")}")
print(f"Title: {data.get("title")}")
```
